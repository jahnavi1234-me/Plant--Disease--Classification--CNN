```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image

DATA_PATH =r"C:\Users\DELL\OneDrive\Desktop\CNN_PROJECT\data\raw"

print(os.listdir(DATA_PATH))

for cls in os.listdir(DATA_PATH):
    print(cls, len(os.listdir(os.path.join(DATA_PATH, cls))))

cls = random.choice(os.listdir(DATA_PATH))
img_name = random.choice(os.listdir(os.path.join(DATA_PATH, cls)))

img = Image.open(os.path.join(DATA_PATH, cls, img_name))
plt.imshow(img)
plt.title(cls)
plt.axis("off")
```
